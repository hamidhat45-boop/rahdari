```python
types = ["گریدر", "بلدوزر", "لودر", "بیل مکانیکی"]

# 1. Filter the data
Heavy_machine = merged_rapier_data_distanse[ merged_rapier_data_distanse["Car Type"].isin(types) ]

# 2. Group by 'Unit Group_x' and aggregate
result = Heavy_machine.groupby(["Unit Group_x", "Car Type"]).agg({
    "Wheels Price": ["sum", "mean"],
    "Fuel Price": ["sum", "mean"],
    "Repair Price": ["sum", "mean"],
    "Service Price": ["sum", "mean"],
    "Overhal Price": ["sum", "mean"],
    "Necs Price": ["sum", "mean"],
    "On time (minutes)": ["sum", "mean"]
})

# 3. Flatten column names
result.columns = [
    "Wheels Price Sum", "Wheels Price Mean",
    "Fuel Price Sum", "Fuel Price Mean",
    "Repair Price Sum", "Repair Price Mean",
    "Service Price Sum", "Service Price Mean",
    "Overhal Price Sum", "Overhal Price Mean",
    "Necs Price Sum", "Necs Price Mean",
    "On time Sum", "On time Mean"
]

# 4. Export to Excel
result.to_excel("Heavy_machine_summary.xlsx", index=True)
```

```python
# List of heavy machinery types
types = ["گریدر", "بلدوزر", "لودر", "بیل مکانیکی"]

# Dictionary to store results for each type
results_dict = {}

for car_type in types:
    # Filter for current car type
    df_type = merged_rapier_data_distanse[merged_rapier_data_distanse["Car Type"] == car_type]
    
    # Group by 'Unit Group_x' and aggregate
    result1 = df_type.groupby("Unit Group_x").agg({
        "Wheels Price": ["sum", "mean"],
        "Fuel Price": ["sum", "mean"],
        "Repair Price": ["sum", "mean"],
        "Service Price": ["sum", "mean"],
        "Overhal Price": ["sum", "mean"],
        "Necs Price": ["sum", "mean"],
        "On time (minutes)": ["sum", "mean"]
    })
    
    # Flatten MultiIndex columns
    result1.columns = [f"{col[0]} {col[1]}" for col in result1.columns]
    
    # Store in dictionary
    results_dict[car_type] = result1

# Export each table to a separate sheet in one Excel file
with pd.ExcelWriter("Heavy_Machine_Summary_by_Type.xlsx") as writer:
    for car_type, df in results_dict.items():
        df.to_excel(writer, sheet_name=car_type)

print("✅ Excel file created: Heavy_Machine_Summary_by_Type.xlsx")
```
